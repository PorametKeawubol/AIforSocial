```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")
for token in doc:
    print(token.text, token.pos_, token.dep_)
```

```python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")

for ent in doc.ents:
    print(ent.text, ent.start_char, ent.end_char, ent.label_)

displacy.serve(doc, style="ent")
```

```python
import re

# Define the text to search in
text = """
    การที่ผู้ร้องยื่นคำร้องขอให้บังคับตามคำชี้ขาดของอนุญาโตตุลาการตาม พรบ.อนุญาโตตุลาการ พ.ศ. 2545 มาตรา 42
    ซึ่งในการยื่นคำร้องดังกล่าวมีประเด็นที่ต้องพิจารณาเพียงว่าศาลจะบังคับตามคำชี้ขาดของอนุญาโตตุลาการหรือไม่เท่านั้น
    วันที่ 25 มกราคม 2564 เป็นต้นไป
    พรบ.อนุญาโตตุลาการ พ.ศ. 2545 มาตรา 42 นี้เป็นต้น
    วันที่ 10 กันยายน 2563 นี้เช่นกัน
    และอีกบรรดาข้อความที่เกี่ยวข้องกับพรบ.อนุญาโตตุลาการ พ.ศ. 2545 มาตรา 42 ที่ถูกกล่าวถึง
"""

# Define the regular expression patterns in an array
patterns = [
    r'\d{1,2}\s+(?:มกราคม|กุมภาพันธ์|มีนาคม|เมษายน|พฤษภาคม|มิถุนายน|กรกฎาคม|สิงหาคม|กันยายน|ตุลาคม|พฤศจิกายน|ธันวาคม)\s+\d{4}',  # วันที่ เดือน ปี
    r'พรบ\.อนุญาโตตุลาการ\s+พ\.ศ\.\s+\d{4}\s+มาตรา\s+\d{2,3}',  # พรบ.อนุญาโตตุลาการ พ.ศ. 2545 มาตรา 42
    r'ผู้คัดค้านที่\s+\d{1}', #ผู้คัดค้านที่ 3
    r'\d{1,2}/\d{1,2}/\d{4}',  # วัน/เดือน/ปี
    r'\d{4}-\d{2}-\d{2}'  # ปี-เดือน-วัน (ISO format)
]

# Initialize a list to store all matches
all_matches = []

# Iterate over each pattern and find all occurrences in the text
for pattern in patterns:
    matches = re.findall(pattern, text2)
    all_matches.extend(matches)

# Print all matches found
if all_matches:
    for match in all_matches:
        print(f"Found: {match}")
else:
    print("No matches found.")
```
